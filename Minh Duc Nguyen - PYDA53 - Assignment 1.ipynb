### Task 1
Trong python, kiểu dữ liệu float quy định phân cách phần thập phân bằng dấu ".", hãy thực hiện convert đưa giá trị bị sai định dạng về đúng kiểu float.

Gợi ý: các bạn vui lòng đọc trước các phương thức của string (string methods), ở đây cần dùng một vài thao tác xử lý như str.replace() để...

```python
invalid_value = ' 23.345,21'  # 23345.21
```

```python
# Start here

invalid_value = ' 23.345,21' # Đề bài yêu cầu phải xuất ra 23345.21

correct_value = invalid_value.replace(".","").strip() # Để bỏ dấu chấm thành "" " 23.345,21" => " 23345,21" và bỏ khoảng trắng ban đầu đề cho " 23..." => "23..."

correct_value = float(correct_value.replace(",",".")) # Để bỏ dấu phẩy thành dấu chấm " 23345,21" => " 23345.21" và chuyển đổi biến về float

print(correct_value) # Chuyển đổi biến về float

# Em ban đầu làm theo correct_value = invalid_value.replace("."," ") rồi correct_value = invalid_value.replace(",",".")
# Nhưng em lại quên là sau khi chuyển . thành "" thì mình phải ghi trên chính cái kết quả mình vừa ra chứ không phải là invalid_value ban đầu
```

### Task 2
Cho trước một xâu ký tự hãy tìm cách lấy ra đoạn ký tự là sđt bằng các cách khác nhau.

Input: 'nguyen tuan Anh DA11 098 844 4888' --> Output mong muốn: '0988444888'

Tương tự task 1, ở đây cũng cần thiết sử dụng một vài thao tác với string như extract xâu con, ...

```python
s = 'nguyen tuan Anh DA11 098 844 4888'  # output can la '0988444888'
```

```python
# Start here

s = 'nguyen tuan Anh DA11 098 844 4888' # Đề bài yêu cầu là '0988444888'

s = s.split("DA11")[-1] # Cắt chuỗi kí tự tại DA11 thành nguyen tuan Anh và 098 844 4888 và lấy phần 098 844 4888

s = s.replace(" ","") # Bỏ khoảng trắng trong chuỗi

print(s)

# Ban đầu em làm là s = s[21:] nhưng cách này nếu thêm 1 kí tự trong xâu chuỗi sẽ sai nên em đã tìm cách khác phù hợp hơn.
```
